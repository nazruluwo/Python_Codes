Write a Python program to find the factorial of a number.

```python
def factorial(n):
  if n == 0 or n == 1:
    return 1;
  return n*factorial(n-1)

print(factorial(int(input('Give a number for calculating factorial'))))
```

Write a Python program to find the largest element in a list.

```python
def largest_numbers(number_list):
    largest = number_list[0]
    for num in number_list:
        if largest < num:
            largest = num
    return largest

print(largest_numbers([32,3,44,5]))
        
```

Write a Python program to reverse a string.

```python
def reverse_string(input_str):
    return input_str[::-1]

print(reverse_string("HELLO WORLD"))
```

Write a Python program to count the frequency of each element in a list.

```python
def frequency_count(num_list):
    frequency = {}
    for num in num_list:
        if num in frequency:
            frequency[num]+=1
        else:
            frequency[num]=1
    return frequency

freq_count = frequency_count([22,33,22,5,66,33,33])

print(freq_count[33])
            
```

Write a Python program to check if a number is prime.

```python
import math

def is_prime(num):
    if num < 2:
        return False
    n = math.sqrt(num) + 1
    for i in range(n):
      if i!=1 and i!=num:
          if num%i==0:
              return False
    return True

num = input('Give a number to check if its prime: ')
if(is_prime(float(num))):
  print(num, 'is a prime')
else:
    print(num, 'is not a prime')

          
```

Write a Python program to find the second largest number in a list.

```python
def find_second_largest(list):
    sort_list = list
    sort_list.sort(reverse=True)
    return sort_list[1]

sl = find_second_largest([2,6,9,5])
print(sl)
```

Write a Python program to remove duplicates from a list.

```python
def remove_duplicate(list):
    unique_list=[]
    for i in list:
        if i not in unique_list:
            unique_list.append(i)
    return unique_list

print(remove_duplicate([1,22,22,22,3,3,4,5,6]))
```
